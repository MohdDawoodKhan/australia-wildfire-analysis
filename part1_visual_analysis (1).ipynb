## 🔥 Task 1.1: Line Chart for Average Fire Area Over Time
Use pandas to group your data by year and calculate the average estimated fire area for each year. Then, use a line chart to show how the average fire area changes over time.

```
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

# Load data
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Historical_Wildfires.csv'
df = pd.read_csv(url)
df['Year'] = pd.to_datetime(df['Date']).dt.year

# Group by Year and plot
df_new = df.groupby('Year')['Estimated_fire_area'].mean()
plt.figure(figsize=(12, 6))
df_new.plot()
plt.xlabel('Year')
plt.ylabel('Average Estimated Fire Area (km²)')
plt.title('Estimated Fire Area over Time')
plt.grid(True)
plt.show()
```

## 🔥 Task 1.2: Fire Area by Year and Month
Now we’ll group by both Year and Month to see monthly trends across years.

```
df['Month'] = pd.to_datetime(df['Date']).dt.month
df_month = df.groupby(['Year', 'Month'])['Estimated_fire_area'].mean()
df_month.plot(figsize=(12, 6))
plt.xlabel('Year, Month')
plt.ylabel('Average Estimated Fire Area (km²)')
plt.title('Monthly Estimated Fire Area Over Years')
plt.grid(True)
plt.show()
```

## 🔥 Task 1.3: Fire Brightness by Region (Barplot)
Compare fire brightness across different Australian regions using Seaborn barplot.

```
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Region', y='Mean_estimated_fire_brightness')
plt.xlabel('Region')
plt.ylabel('Mean Estimated Fire Brightness (Kelvin)')
plt.title('Distribution of Mean Estimated Fire Brightness across Regions')
plt.show()
```

## 🔥 Task 1.4: Pie Chart of Pixel Count by Region
Show the portion of fire pixel counts across different regions.

```
plt.figure(figsize=(10, 6))
region_counts = df.groupby('Region')['Count'].sum()
plt.pie(region_counts, labels=region_counts.index, autopct='%1.1f%%')
plt.title('Percentage of Pixels for Presumed Vegetation Fires by Region')
plt.axis('equal')
plt.show()
```

## 🔥 Task 1.5: Improved Pie Chart with Legend
Customize the pie chart with a better label format using `plt.legend()`.

```
plt.figure(figsize=(10, 6))
region_counts = df.groupby('Region')['Count'].sum()
plt.pie(region_counts, labels=region_counts.index)
plt.legend([(i, round(k / region_counts.sum() * 100, 2)) for i, k in zip(region_counts.index, region_counts)])
plt.title('Percentage of Pixels for Presumed Vegetation Fires by Region')
plt.axis('equal')
plt.show()
```

## 🔥 Task 1.6: Histogram of Fire Brightness
Show the distribution of brightness values using a histogram.

```
plt.figure(figsize=(10, 6))
plt.hist(x=df['Mean_estimated_fire_brightness'], bins=20)
plt.xlabel('Mean Estimated Fire Brightness (Kelvin)')
plt.ylabel('Count')
plt.title('Histogram of Mean Estimated Fire Brightness')
plt.show()
```

## 🔥 Task 1.7: Brightness Distribution by Region
Use `hue='Region'` in Seaborn to show regional distribution in one plot.

```
plt.figure(figsize=(12, 6))
sns.histplot(data=df, x='Mean_estimated_fire_brightness', hue='Region', multiple='stack')
plt.title('Brightness Distribution by Region')
plt.xlabel('Mean Estimated Fire Brightness (Kelvin)')
plt.ylabel('Count')
plt.grid(True)
plt.show()
```

## 🔥 Task 1.8: Scatter Plot - Radiative Power vs Confidence
Visualize the relationship between fire radiative power and detection confidence.

```
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Mean_confidence', y='Mean_estimated_fire_radiative_power')
plt.xlabel('Mean Confidence')
plt.ylabel('Mean Estimated Fire Radiative Power (MW)')
plt.title('Radiative Power vs. Confidence')
plt.grid(True)
plt.show()
```

## 🔥 Task 1.9: Map of Regions Using Folium
Use latitude and longitude to map regions in Australia.

```
import folium
region_data = {
    'region': ['NSW', 'QL', 'SA', 'TA', 'VI', 'WA', 'NT'],
    'Lat': [-31.87, -22.16, -30.53, -42.03, -36.59, -25.23, -19.49],
    'Lon': [147.28, 144.58, 135.63, 146.63, 144.67, 121.01, 132.55]
}
import pandas as pd
reg = pd.DataFrame(region_data)
aus_reg = folium.map.FeatureGroup()
Aus_map = folium.Map(location=[-25, 135], zoom_start=4)
for lat, lng, lab in zip(reg.Lat, reg.Lon, reg.region):
    aus_reg.add_child(folium.CircleMarker(
        location=[lat, lng],
        popup=lab,
        radius=5,
        color='red',
        fill=True,
        fill_color='blue',
        fill_opacity=0.6
    ))
Aus_map.add_child(aus_reg)
Aus_map
```

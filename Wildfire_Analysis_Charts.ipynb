## 📊 Task 1.1: Line chart for average fire area over time
**Goal:** Analyze how the average fire area changes year by year.

```

import pandas as pd
import matplotlib.pyplot as plt

# Load data
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Historical_Wildfires.csv"
df = pd.read_csv(url)
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year

# Plot average fire area per year
plt.figure(figsize=(12, 6))
df_new = df.groupby('Year')['Estimated_fire_area'].mean()
df_new.plot()
plt.xlabel('Year')
plt.ylabel('Average Estimated Fire Area (km²)')
plt.title('Estimated Fire Area over Time')
plt.grid(True)
plt.savefig("fire_area_trend.png")
plt.show()
```

## 📊 Task 1.2: Estimated fire area by month
**Goal:** Identify months with highest fire activity.

```

df['Month'] = df['Date'].dt.month
df_month = df.groupby(['Year', 'Month'])['Estimated_fire_area'].mean()
df_month.plot(figsize=(12, 6))
plt.xlabel('Year, Month')
plt.ylabel('Average Estimated Fire Area (km²)')
plt.title('Estimated Fire Area per Month over Time')
plt.grid(True)
plt.savefig("fire_area_monthly.png")
plt.show()
```

## 📊 Task 1.3: Seaborn barplot for brightness across regions

```

import seaborn as sns

plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Region', y='Mean_estimated_fire_brightness')
plt.xlabel('Region')
plt.ylabel('Mean Estimated Fire Brightness (Kelvin)')
plt.title('Distribution of Mean Estimated Fire Brightness across Regions')
plt.savefig("brightness_by_region.png")
plt.show()
```

## 📊 Task 1.4 & 1.5: Pie chart of pixel count per region and customization

```

plt.figure(figsize=(10, 6))
region_counts = df.groupby('Region')['Count'].sum()
plt.pie(region_counts, labels=region_counts.index)
plt.legend([(i, round(k/region_counts.sum()*100, 2)) for i,k in zip(region_counts.index, region_counts)])
plt.title('Percentage of Pixels for Presumed Vegetation Fires by Region')
plt.axis('equal')
plt.savefig("pixels_pie_chart.png")
plt.show()
```

## 📊 Task 1.6: Histogram of fire brightness

```

plt.figure(figsize=(10, 6))
plt.hist(x=df['Mean_estimated_fire_brightness'], bins=20)
plt.xlabel('Mean Estimated Fire Brightness (Kelvin)')
plt.ylabel('Count')
plt.title('Histogram of Mean Estimated Fire Brightness')
plt.grid(True)
plt.savefig("brightness_histogram.png")
plt.show()
```

## 📊 Task 1.7: Seaborn distribution by region (stacked histogram)

```

sns.histplot(data=df, x='Mean_estimated_fire_brightness', hue='Region', multiple='stack')
plt.title('Stacked Distribution of Mean Estimated Fire Brightness by Region')
plt.xlabel('Mean Estimated Fire Brightness (Kelvin)')
plt.ylabel('Frequency')
plt.savefig("brightness_by_region_stacked.png")
plt.show()
```

## 📊 Task 1.8: Scatter plot of radiative power vs confidence

```

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Mean_confidence', y='Mean_estimated_fire_radiative_power')
plt.xlabel('Mean Confidence')
plt.ylabel('Mean Estimated Fire Radiative Power (MW)')
plt.title('Mean Estimated Fire Radiative Power vs. Mean Confidence')
plt.grid(True)
plt.savefig("radiative_vs_confidence.png")
plt.show()
```

## 📊 Task 1.9: Map regions using Folium

```

import folium

region_data = {
    'region': ['NSW','QL','SA','TA','VI','WA','NT'],
    'Lat': [-31.87, -22.16, -30.53, -42.03, -36.59, -25.23, -19.49],
    'Lon': [147.28, 144.58, 135.63, 146.63, 144.67, 121.01, 132.55]
}

reg = pd.DataFrame(region_data)
aus_map = folium.Map(location=[-25, 135], zoom_start=4)
for lat, lon, name in zip(reg['Lat'], reg['Lon'], reg['region']):
    folium.CircleMarker(location=[lat, lon], popup=name, radius=5, color='red', fill=True).add_to(aus_map)

aus_map.save("australia_wildfire_map.html")
aus_map
```
